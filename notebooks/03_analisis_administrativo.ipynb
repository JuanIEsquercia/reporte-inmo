# Análisis del Módulo Administrativo

Este notebook analiza la estructura del archivo Excel del módulo administrativo para procesar y analizar la información de consultas/leads.


## 1. Importación de Librerías


```python
import pandas as pd
import numpy as np
import os

# Configuración para mostrar todas las columnas
pd.set_option('display.max_columns', None)
pd.set_option('display.max_rows', 100)
pd.set_option('display.width', None)
pd.set_option('display.max_colwidth', 50)
```

## 2. Carga del archivo Excel


```python
# Ruta del archivo
archivo_excel = '../data/administrativo/kommo_export_leads_2025-12-01.xlsx'

# Cargar el archivo Excel
print("Cargando archivo Excel...")
print(f"Archivo: {archivo_excel}")

# Primero, verificar qué hojas tiene el archivo
xl_file = pd.ExcelFile(archivo_excel)
print(f"\nHojas disponibles en el archivo:")
for sheet in xl_file.sheet_names:
    print(f"  - {sheet}")

# Cargar todas las hojas
dataframes = {}
for sheet in xl_file.sheet_names:
    print(f"\nCargando hoja: {sheet}")
    df = pd.read_excel(archivo_excel, sheet_name=sheet)
    dataframes[sheet] = df
    print(f"  Filas: {len(df)}, Columnas: {len(df.columns)}")
    print(f"  Columnas: {list(df.columns)}")
```

## 3. Exploración de la estructura de datos


```python
# Explorar cada hoja en detalle
for nombre_hoja, df in dataframes.items():
    print(f"\n{'='*80}")
    print(f"HOJA: {nombre_hoja}")
    print(f"{'='*80}")
    print(f"\nDimensiones: {df.shape[0]} filas x {df.shape[1]} columnas")
    print(f"\nPrimeras filas:")
    print(df.head(10))
    print(f"\nInformación de tipos de datos:")
    print(df.dtypes)
    print(f"\nValores nulos por columna:")
    print(df.isnull().sum())
    print(f"\nMuestra de datos únicos por columna (primeras 5 columnas):")
    for col in df.columns[:5]:
        valores_unicos = df[col].unique()[:10]
        print(f"\n  {col}: {valores_unicos}")
```

## 4. Análisis de datos vacíos y limpieza inicial


```python
# Identificar filas completamente vacías
for nombre_hoja, df in dataframes.items():
    print(f"\n{nombre_hoja}:")
    filas_vacias = df.isnull().all(axis=1).sum()
    print(f"  Filas completamente vacías: {filas_vacias}")
    
    # Mostrar ejemplo de filas vacías
    if filas_vacias > 0:
        print(f"  Ejemplo de filas vacías (índices):")
        indices_vacios = df[df.isnull().all(axis=1)].index[:5]
        print(f"    {list(indices_vacios)}")
```

## 5. Eliminación de columnas no relevantes


```python
# Columnas a eliminar
columnas_eliminar = [
    'Nombre del lead',
    'Compañía',
    'Compañía del lead',
    'Modificado por',
    'Etiquetas del lead',
    'Tareas próximas',
    'Cerrado el',
    'utm_content',
    'utm_medium',
    'utm_campaign',
    'utm_source',
    'utm_term',
    'utm_referrer',
    'referrer',
    'gclientid',
    'gclid',
    'fbclid',
    'Cargo (contacto)',
    'Correo (contacto)',
    'E-mail priv. (contacto)',
    'Otro e-mail (contacto)',
    'Teléfono oficina (contacto)',
    'Teléfono oficina directo (contacto)',
    'Teléfono celular (contacto)',
    'Fax (contacto)',
    'Teléfono de casa (contacto)',
    'Otro teléfono (contacto)',
    'Términos y condiciones (contacto)',
    'Nota 1',
    'Nota 2',
    'Nota 3',
    'Nota 4',
    'Nota 5'
]

# Eliminar columnas de cada hoja
for nombre_hoja, df in dataframes.items():
    print(f"\n{nombre_hoja}:")
    print(f"  Columnas antes: {len(df.columns)}")
    
    # Verificar qué columnas existen antes de eliminar
    columnas_existentes = [col for col in columnas_eliminar if col in df.columns]
    columnas_no_existentes = [col for col in columnas_eliminar if col not in df.columns]
    
    if columnas_no_existentes:
        print(f"  ⚠️ Columnas no encontradas (se ignorarán): {columnas_no_existentes}")
    
    # Eliminar columnas
    df_limpio = df.drop(columns=columnas_existentes, errors='ignore')
    dataframes[nombre_hoja] = df_limpio
    
    print(f"  Columnas después: {len(df_limpio.columns)}")
    print(f"  Columnas eliminadas: {len(columnas_existentes)}")
    print(f"\n  Columnas que quedan:")
    for i, col in enumerate(df_limpio.columns, 1):
        print(f"    {i}. {col}")
```

## 6. Verificación de datos después de la limpieza


```python
# Mostrar resumen de cada hoja después de la limpieza
for nombre_hoja, df in dataframes.items():
    print(f"\n{'='*80}")
    print(f"HOJA: {nombre_hoja} - DESPUÉS DE LIMPIEZA")
    print(f"{'='*80}")
    print(f"\nDimensiones: {df.shape[0]} filas x {df.shape[1]} columnas")
    print(f"\nPrimeras filas:")
    print(df.head(10))
    print(f"\nInformación de tipos de datos:")
    print(df.dtypes)
    print(f"\nValores nulos por columna:")
    print(df.isnull().sum())
```

## 7. Eliminación de filas duplicadas


```python
# Eliminar filas duplicadas basadas en "Contacto principal"
for nombre_hoja, df in dataframes.items():
    print(f"\n{nombre_hoja}:")
    print(f"  Filas antes de eliminar duplicados: {len(df)}")
    
    # Verificar si existe la columna "Contacto principal"
    if 'Contacto principal' in df.columns:
        # Contar duplicados antes de eliminar
        duplicados = df.duplicated(subset=['Contacto principal'], keep=False)
        cantidad_duplicados = duplicados.sum()
        print(f"  Filas duplicadas encontradas: {cantidad_duplicados}")
        
        if cantidad_duplicados > 0:
            print(f"\n  Ejemplos de contactos duplicados:")
            contactos_duplicados = df[df.duplicated(subset=['Contacto principal'], keep=False)]['Contacto principal'].value_counts().head(10)
            for contacto, count in contactos_duplicados.items():
                print(f"    - {contacto}: {count} veces")
        
        # Eliminar duplicados, manteniendo la primera ocurrencia
        df_sin_duplicados = df.drop_duplicates(subset=['Contacto principal'], keep='first')
        dataframes[nombre_hoja] = df_sin_duplicados
        
        print(f"\n  Filas después de eliminar duplicados: {len(df_sin_duplicados)}")
        print(f"  Filas eliminadas: {len(df) - len(df_sin_duplicados)}")
    else:
        print(f"  ⚠️ No se encontró la columna 'Contacto principal'")
```

## 8. Verificación final de datos limpios


```python
# Mostrar resumen final de cada hoja
for nombre_hoja, df in dataframes.items():
    print(f"\n{'='*80}")
    print(f"HOJA: {nombre_hoja} - DATOS FINALES LIMPIOS")
    print(f"{'='*80}")
    print(f"\nDimensiones: {df.shape[0]} filas x {df.shape[1]} columnas")
    print(f"\nColumnas finales:")
    for i, col in enumerate(df.columns, 1):
        print(f"  {i}. {col}")
    print(f"\nPrimeras 10 filas:")
    print(df.head(10))
    print(f"\nValores nulos por columna:")
    print(df.isnull().sum())
```

## 9. Análisis y KPIs del Módulo Administrativo


```python
# Obtener el dataframe principal (Sheet1)
df_administrativo = dataframes.get('Sheet1', None)

if df_administrativo is not None:
    print("="*80)
    print("ANÁLISIS Y KPIs DEL MÓDULO ADMINISTRATIVO")
    print("="*80)
    
    # KPI 1: Total de Contactos Principales
    print("\n" + "="*80)
    print("KPI 1: Total de Contactos Principales")
    print("="*80)
    total_contactos = len(df_administrativo)
    print(f"Total de contactos principales: {total_contactos}")
    
    # KPI 2: Distribución por Responsable
    print("\n" + "="*80)
    print("KPI 2: Distribución por Responsable")
    print("="*80)
    if 'Responsable' in df_administrativo.columns:
        distribucion_responsable = df_administrativo['Responsable'].value_counts()
        print("\nCantidad de contactos por Responsable:")
        for responsable, cantidad in distribucion_responsable.items():
            porcentaje = (cantidad / total_contactos) * 100
            print(f"  {responsable}: {cantidad} contactos ({porcentaje:.2f}%)")
    else:
        print("⚠️ No se encontró la columna 'Responsable'")
    
    # KPI 3: Análisis de Fecha de Creación
    print("\n" + "="*80)
    print("KPI 3: Análisis de Fecha de Creación")
    print("="*80)
    if 'Fecha de Creación' in df_administrativo.columns:
        # Convertir a datetime
        df_administrativo['Fecha de Creación'] = pd.to_datetime(df_administrativo['Fecha de Creación'], errors='coerce')
        
        # Filtrar solo las fechas válidas
        df_fechas_validas = df_administrativo[df_administrativo['Fecha de Creación'].notna()].copy()
        
        print(f"\nTotal de registros con fecha válida: {len(df_fechas_validas)}")
        print(f"Rango de fechas: {df_fechas_validas['Fecha de Creación'].min()} a {df_fechas_validas['Fecha de Creación'].max()}")
        
        # Crear columnas auxiliares para análisis
        df_fechas_validas['Año'] = df_fechas_validas['Fecha de Creación'].dt.year
        df_fechas_validas['Mes'] = df_fechas_validas['Fecha de Creación'].dt.month
        df_fechas_validas['Año-Mes'] = df_fechas_validas['Fecha de Creación'].dt.to_period('M')
        df_fechas_validas['Semana'] = df_fechas_validas['Fecha de Creación'].dt.to_period('W')
        
        # Acumulado total
        print("\n--- Acumulado Total ---")
        df_fechas_validas_sorted = df_fechas_validas.sort_values('Fecha de Creación')
        df_fechas_validas_sorted['Acumulado'] = range(1, len(df_fechas_validas_sorted) + 1)
        print(f"Total acumulado: {len(df_fechas_validas_sorted)} contactos")
        
        # Por mes
        print("\n--- Distribución por Mes ---")
        conteo_mes = df_fechas_validas.groupby('Año-Mes').size().sort_index()
        acumulado_mes = conteo_mes.cumsum()
        print("\nContactos creados por mes:")
        for periodo, cantidad in conteo_mes.items():
            acumulado = acumulado_mes[periodo]
            print(f"  {periodo}: {cantidad} contactos (Acumulado: {acumulado})")
        
        # Por semana
        print("\n--- Distribución por Semana ---")
        conteo_semana = df_fechas_validas.groupby('Semana').size().sort_index()
        acumulado_semana = conteo_semana.cumsum()
        print("\nContactos creados por semana (últimas 10 semanas):")
        for periodo, cantidad in list(conteo_semana.items())[-10:]:
            acumulado = acumulado_semana[periodo]
            print(f"  {periodo}: {cantidad} contactos (Acumulado: {acumulado})")
        
    else:
        print("⚠️ No se encontró la columna 'Fecha de Creación'")
else:
    print("⚠️ No se encontró el dataframe 'Sheet1'")
```

```python
# Crear tablas resumen para facilitar el análisis
if df_administrativo is not None and 'Fecha de Creación' in df_administrativo.columns:
    df_fechas_validas = df_administrativo[df_administrativo['Fecha de Creación'].notna()].copy()
    
    # Función para parsear fechas (ya debería estar parseada, pero por si acaso)
    def parsear_fecha(fecha):
        """Parsea fechas en formato día/mes/año (formato latino)"""
        if pd.isna(fecha):
            return pd.NaT
        if isinstance(fecha, pd.Timestamp):
            return fecha
        fecha_str = str(fecha).strip()
        formatos = ['%d/%m/%Y', '%d-%m-%Y', '%Y-%m-%d', '%d/%m/%Y %H:%M:%S', '%Y-%m-%d %H:%M:%S']
        for formato in formatos:
            try:
                return pd.to_datetime(fecha_str, format=formato)
            except:
                continue
        try:
            return pd.to_datetime(fecha_str, dayfirst=True, errors='coerce')
        except:
            return pd.NaT
    
    # Solo parsear si no está ya parseada
    if not pd.api.types.is_datetime64_any_dtype(df_fechas_validas['Fecha de Creación']):
        df_fechas_validas['Fecha de Creación'] = df_fechas_validas['Fecha de Creación'].apply(parsear_fecha)
    df_fechas_validas['Año-Mes'] = df_fechas_validas['Fecha de Creación'].dt.to_period('M')
    df_fechas_validas['Semana'] = df_fechas_validas['Fecha de Creación'].dt.to_period('W')
    
    # Tabla resumen por mes
    print("\n" + "="*80)
    print("TABLA RESUMEN: Contactos por Mes")
    print("="*80)
    resumen_mes = df_fechas_validas.groupby('Año-Mes').agg({
        'ID': 'count',
        'Fecha de Creación': ['min', 'max']
    }).reset_index()
    resumen_mes.columns = ['Año-Mes', 'Cantidad', 'Fecha Mínima', 'Fecha Máxima']
    resumen_mes['Acumulado'] = resumen_mes['Cantidad'].cumsum()
    print(resumen_mes.to_string(index=False))
    
    # Tabla resumen por semana
    print("\n" + "="*80)
    print("TABLA RESUMEN: Contactos por Semana (últimas 15 semanas)")
    print("="*80)
    resumen_semana = df_fechas_validas.groupby('Semana').agg({
        'ID': 'count',
        'Fecha de Creación': ['min', 'max']
    }).reset_index()
    resumen_semana.columns = ['Semana', 'Cantidad', 'Fecha Mínima', 'Fecha Máxima']
    resumen_semana['Acumulado'] = resumen_semana['Cantidad'].cumsum()
    print(resumen_semana.tail(15).to_string(index=False))
    
    # Tabla resumen por Responsable
    print("\n" + "="*80)
    print("TABLA RESUMEN: Contactos por Responsable")
    print("="*80)
    if 'Responsable' in df_administrativo.columns:
        resumen_responsable = df_administrativo['Responsable'].value_counts().reset_index()
        resumen_responsable.columns = ['Responsable', 'Cantidad']
        resumen_responsable['Porcentaje'] = (resumen_responsable['Cantidad'] / len(df_administrativo) * 100).round(2)
        print(resumen_responsable.to_string(index=False))
```
